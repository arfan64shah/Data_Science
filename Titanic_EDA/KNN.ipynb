```python
# import all the necessary libraries
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from joblib import dump
```

```python
# now import the dataset and read it
dataset = pd.read_csv('final_dataset.csv')
dataset.head()
```

```python
# even though the dataset is cleaned but lets verify it again quickly
dataset.isnull().sum()
```

```python
# also search for data types
dataset.dtypes
```

```python
# seems the dataset is ready to split it into dependent and independent or x and y
x = dataset.iloc[:, :-1]
y = dataset.iloc[:, -1]
y
```

```python
# apply standard scalling method
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
# lets now split x and y into training and testing

# chose a spliting point
split_index = int(len(dataset)*0.85)
x_train = x[:split_index, :]
x_test = x[split_index:, :]

y_train = y[:split_index]
y_test = y[split_index:]
```

```python
# print shape of training and testing
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
# now build the model
model = KNeighborsClassifier(n_neighbors = 12)

# fit the data into model
model.fit(x_train, y_train)
```

```python
# predict the output
y_pred = model.predict(x_test)
```

```python
# now check the accuracy
score = accuracy_score(y_pred, y_test)
score
```

```python
# save the model
dump(model, 'knn_titanic_model.joblib')
```
